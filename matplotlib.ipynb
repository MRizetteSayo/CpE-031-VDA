Markers
- You can use the argument marker to emphasize each point with a specified marker

```python
#plotting x and y points
ypoints=np.array([3,8,1,10])

plt.plot(ypoints,marker='o')  #plot() is used to draw points
plt.show()
```

```python
#plotting x and y points
ypoints=np.array([3,8,1,10])

plt.plot(ypoints,marker='x')  #plot() is used to draw points
plt.show()
```

Format Strings fmt

```python
#plotting x and y points
ypoints=np.array([3,8,1,10])

plt.plot(ypoints,'o:b')  #marker | line | color
plt.show()
```

Marker Size

```python
#plotting x and y points
ypoints=np.array([3,8,1,10])

plt.plot(ypoints,'o:b',ms=20)  #marker | line | color
plt.show()
```

Marker Color
-You can use the argument markeredgecolor or mec for short

```python
#plotting x and y points
ypoints=np.array([3,8,1,10])

plt.plot(ypoints,'o:b',ms=10,mec='r')  #marker | line | color
plt.show()
```

Marker Face Color or mfc

```python
#plotting x and y points
ypoints=np.array([3,8,1,10])

plt.plot(ypoints,'o:b',ms=10,mec='green',mfc='red')  #marker | line | color
plt.show()
```

```python
plt.plot(ypoints,linestyle='dashed')
```

Matplot Labels and Title
-You can use the xlabel() and ylabel()

```python
xpoints=np.array([1,2,6,8])  #plotting x and y points
ypoints=np.array([3,8,1,10])

plt.plot(xpoints,ypoints)  #plot() is used to draw points

plt.xlabel("Distance in km")
plt.ylabel("Speed in km/s")
plt.show()
```

```python
xpoints=np.array([1,2,6,8])  #plotting x and y points
ypoints=np.array([3,8,1,10])

plt.plot(xpoints,ypoints)  #plot() is used to draw points

plt.title("Distance Travelled Everyday")
plt.xlabel("Distance in km") #labels
plt.ylabel("Speed in km/s")
plt.show()
```

```python
plt.title("Distance Travelled Everyday", loc='left')
plt.xlabel("Distance in km") #labels
plt.ylabel("Speed in km/s")
plt.show()
```

Matplot Grid
-You can use grid() to add grid lines

```python
xpoints=np.array([1,2,6,8])  #plotting x and y points
ypoints=np.array([3,8,1,10])

plt.plot(xpoints,ypoints)  #plot() is used to draw points
plt.title("Distance Travelled Everyday")
plt.xlabel("Distance in km") #labels
plt.ylabel("Speed in km/s")
plt.grid()
plt.show()
```

```python
#select grid lines to display
xpoints=np.array([1,2,6,8])  #plotting x and y points
ypoints=np.array([3,8,1,10])

plt.plot(xpoints,ypoints)  #plot() is used to draw points
plt.title("Distance Travelled Everyday")
plt.xlabel("Distance in km") #labels
plt.ylabel("Speed in km/s")
plt.grid(axis='y')
plt.show()
```

```python
#add properties to grid lines
xpoints=np.array([1,2,6,8])  #plotting x and y points
ypoints=np.array([3,8,1,10])

plt.plot(xpoints,ypoints)  #plot() is used to draw points
plt.title("Distance Travelled Everyday")
plt.xlabel("Distance in km") #labels
plt.ylabel("Speed in km/s")
plt.grid(axis='x',linestyle='--',linewidth=0.50)
plt.show()
```

Show Multiple Plots


```python
#plot 1
xpoints=np.array([1,2,6,8])  #plotting x and y points
ypoints=np.array([3,8,1,10])
plt.subplot(1,2,1)
plt.plot(xpoints,ypoints)  #plot() is used to draw points

#plot2
xpoints=np.array([1,2,6,8])  #plotting x and y points
ypoints=np.array([10,20,30,40])
plt.subplot(1,2,2)
plt.plot(xpoints,ypoints)  #plot() is used to draw points

plt.grid(axis='x',linestyle='--',linewidth=0.50)
plt.show()
```

Matplotlib Bars
- You can use the bar() to draw bars


```python
import matplotlib.pyplot as plt
import numpy as np

x=np.array(['Student1','Student2','Student3'])
y=np.array([90,85,86])
plt.bar(x,y)
plt.show()
```

```python
#Horizontal Bar
plt.barh(x,y)
plt.show()
```

```python
#Change its color
plt.bar(x,y,color='red')
plt.show()
```

Matplot Histogram
- A histogram showing its frequency distribution


```python
import matplotlib.pyplot as plt
import numpy as np

data=[1,2,3,3,3,4,4]
plt.hist(data)
plt.show()
```

```python
#random
import matplotlib.pyplot as plt
import numpy as np
data=np.random.normal(170,10,250)
plt.hist(data)
plt.show()
```
